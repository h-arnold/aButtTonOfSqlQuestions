### Exercise 1: Winter Campaigns With 89-Day Duration (SELECT, WHERE, ORDER BY)
Write a query to output `campaign_id`, `name`, `duration_days`, and `total_budget` for campaigns where `duration_days` equals `89`.
Order by `total_budget` descending.
Skill focus: exact-value filtering on `INTEGER` and ordering by `REAL`.



```python
%%sql
SELECT campaign_id, name, duration_days, total_budget
FROM campaigns
WHERE duration_days = 89
ORDER BY total_budget DESC;
```

### Exercise 2: Long Campaigns With Lower Budgets (SELECT, WHERE, AND, ORDER BY)
Write a query to show `campaign_id`, `name`, `duration_days`, and `total_budget` for campaigns where `duration_days` is at least `82` and `total_budget` is below `30000`.
Order by `duration_days` descending, then `total_budget` ascending.
Skill focus: compound filtering with mixed numeric data types.



```python
%%sql
SELECT campaign_id, name, duration_days, total_budget
FROM campaigns
WHERE duration_days >= 82 AND total_budget < 30000
ORDER BY duration_days DESC, total_budget ASC;
```

### Exercise 3: Instagram Video Ads For All Genders (SELECT, WHERE, AND, ORDER BY)
Write a query to output `ad_id`, `campaign_id`, `ad_platform`, `ad_type`, and `target_gender` for ads where platform is `Instagram`, ad type is `Video`, and target gender is `All`.
Order by `ad_id` ascending.
Skill focus: multi-condition filtering on `TEXT` fields.



```python
%%sql
SELECT ad_id, campaign_id, ad_platform, ad_type, target_gender
FROM ads
WHERE ad_platform = 'Instagram' AND ad_type = 'Video' AND target_gender = 'All'
ORDER BY ad_id ASC;
```

### Exercise 4: Narrow Senior User Segment (SELECT, WHERE, AND, IN, ORDER BY)
Write a query to show `user_id`, `country`, `user_gender`, and `age_group` for users who are in age group `55-65`, are `Female`, and whose country is in `France`, `Japan`, or `Mexico`.
Order by `country` ascending, then `user_id` ascending.
Skill focus: targeted segmentation using `IN` with additional filters.



```python
%%sql
SELECT user_id, country, user_gender, age_group
FROM users
WHERE age_group = '55-65' AND user_gender = 'Female' AND country IN ('France', 'Japan', 'Mexico')
ORDER BY country ASC, user_id ASC;
```

### Exercise 5: Click Events By Time Of Day (SELECT, WHERE, GROUP BY, ORDER BY)
Write a query to output `time_of_day` and the number of click events from `ad_events` where `event_type` is `Click`.
Group by `time_of_day` and order by the count descending.
Skill focus: grouped aggregation over categorical event data.



```python
%%sql
SELECT time_of_day, COUNT(*)
FROM ad_events
WHERE event_type = 'Click'
GROUP BY time_of_day
ORDER BY COUNT(*) DESC;
```

### Exercise 6: Ads In The Shortest Campaign (Subquery)
Use a subquery to output `ad_id`, `ad_platform`, and `ad_type` for ads that belong to the campaign with the minimum `duration_days`.
Order by `ad_id` ascending.
Skill focus: nested SELECT to link one table filter to another table result.



```python
%%sql
SELECT ad_id, ad_platform, ad_type
FROM ads
WHERE campaign_id = (
    SELECT campaign_id
    FROM campaigns
    WHERE duration_days = (SELECT MIN(duration_days) FROM campaigns)
)
ORDER BY ad_id ASC;
```

### Exercise 7: Instagram Purchase Events By Ad Type (JOIN, WHERE, GROUP BY)
Use a join between `ad_events` and `ads` to output each `ad_type` and the number of purchase events where event type is `Purchase` and ad platform is `Instagram`.
Group by `ad_type` and order by the purchase count descending.
Skill focus: relational aggregation across fact and dimension tables.



```python
%%sql
SELECT ads.ad_type, COUNT(*)
FROM ad_events
JOIN ads ON ad_events.ad_id = ads.ad_id
WHERE ad_events.event_type = 'Purchase' AND ads.ad_platform = 'Instagram'
GROUP BY ads.ad_type
ORDER BY COUNT(*) DESC;
```
