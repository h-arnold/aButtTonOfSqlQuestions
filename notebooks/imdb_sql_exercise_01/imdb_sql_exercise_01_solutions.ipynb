### Exercise 1: High Revenue Movies (SELECT, WHERE, ORDER BY)
Write a query to show the `title` and `revenue` for movies where revenue is at least `1500000000`.
Order the output by `revenue` in descending order.


```python
%%sql
SELECT title, revenue
FROM movies
WHERE revenue >= 1500000000
ORDER BY revenue DESC;
```

### Exercise 2: Popularity Band (SELECT, WHERE, BETWEEN)
Write a query to display `title` and `vote_count` for movies with `vote_count` between `11500` and `14000` inclusive.
Sort by `vote_count` descending.


```python
%%sql
SELECT title, vote_count
FROM movies
WHERE vote_count BETWEEN 11500 AND 14000
ORDER BY vote_count DESC;
```

### Exercise 3: Highly Rated Blockbusters (SELECT, WHERE, AND, ORDER BY)
Return `title`, `vote_average`, and `vote_count` for movies with `vote_average` at least `8.0` and `vote_count` at least `10000`.
Order by `vote_average` descending, then `vote_count` descending.


```python
%%sql
SELECT title, vote_average, vote_count
FROM movies
WHERE vote_average >= 8.0 AND vote_count >= 10000
ORDER BY vote_average DESC, vote_count DESC;
```

### Exercise 4: Movies By One Director (Subquery)
Use a subquery to list `title` and `release_date` for movies directed by `Colin Trevorrow`.
Order by `release_date` ascending.


```python
%%sql
SELECT title, release_date
FROM movies
WHERE director_id = (
    SELECT id
    FROM directors
    WHERE name = 'Colin Trevorrow'
)
ORDER BY release_date;
```

### Exercise 5: Movie and Director Pairing (JOIN)
Use a join to output movie `title` and director `name` for movies directed by `Shane Black`.
Order by movie `revenue` descending.


```python
%%sql
SELECT m.title, d.name AS director_name
FROM movies AS m
JOIN directors AS d ON m.director_id = d.id
WHERE d.name = 'Shane Black'
ORDER BY m.revenue DESC;
```

### Exercise 6: Selected Directors Using IN (Subquery + IN)
Show `title` and `release_date` for movies directed by any of: `Colin Trevorrow`, `Shane Black`, `Jennifer Yuh Nelson`.
Use `IN` with a subquery and order by `release_date`.


```python
%%sql
SELECT title, release_date
FROM movies
WHERE director_id IN (
    SELECT id
    FROM directors
    WHERE name IN ('Colin Trevorrow', 'Shane Black', 'Jennifer Yuh Nelson')
)
ORDER BY release_date;
```

### Exercise 7: Movie Totals Per Director (JOIN + GROUP BY)
For director IDs `4783, 4784, 4853, 4877, 4879, 4895`, output director `name` and number of movies as `movie_count`.
Group correctly and order by `movie_count` descending then `name` ascending.


```python
%%sql
SELECT d.name, COUNT(*) AS movie_count
FROM directors AS d
JOIN movies AS m ON d.id = m.director_id
WHERE d.id IN (4783, 4784, 4853, 4877, 4879, 4895)
GROUP BY d.id, d.name
ORDER BY movie_count DESC, d.name;
```

### Exercise 8: Create A New Table (CREATE TABLE)
Create a table named `shortlist` with these fields:
- `shortlist_id` as `INT`, `NOT NULL`, and `PRIMARY KEY`
- `movie_title` as `CHAR(60)` and `NOT NULL`
- `review_score` as `NUMERIC(3,1)` and `NOT NULL`
- `shortlisted_on` as `DATETIME` and `NOT NULL`


```python
%%sql
CREATE TABLE shortlist (
    shortlist_id INT NOT NULL PRIMARY KEY,
    movie_title CHAR(60) NOT NULL,
    review_score NUMERIC(3,1) NOT NULL,
    shortlisted_on DATETIME NOT NULL
);
```
