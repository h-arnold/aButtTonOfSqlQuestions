### Exercise 1: Recent High-Engagement Releases (SELECT, WHERE, AND, ORDER BY)
Show `title` and `release_date` for films released on or after `2016-01-01` with `vote_count` at least `6500`.
Order by `release_date` ascending.


```python
%%sql
SELECT title, release_date
FROM movies
WHERE release_date >= '2016-01-01' AND vote_count >= 6500
ORDER BY release_date;
```

### Exercise 2: Revenue Band Query (SELECT, WHERE, AND, ORDER BY)
Return `title` and `revenue` for films with revenue between `1400000000` and `1700000000` inclusive.
Order by `revenue` descending.


```python
%%sql
SELECT title, revenue
FROM movies
WHERE revenue >= 1400000000 AND revenue <= 1700000000
ORDER BY revenue DESC;
```

### Exercise 3: Pre-1980 Popular Titles (SELECT, WHERE, AND, ORDER BY)
List `title`, `release_date`, and `vote_count` for films released before `1980-01-01` with at least `3000` votes.
Order by `release_date` ascending.


```python
%%sql
SELECT title, release_date, vote_count
FROM movies
WHERE release_date < '1980-01-01' AND vote_count >= 3000
ORDER BY release_date;
```

### Exercise 4: Nolan High-Vote Films (Subquery)
Using a subquery, show `title` and `vote_count` for films directed by `Christopher Nolan` where `vote_count` is at least `8000`.
Order by `vote_count` descending.


```python
%%sql
SELECT title, vote_count
FROM movies
WHERE director_id = (
    SELECT id
    FROM directors
    WHERE name = 'Christopher Nolan'
)
AND vote_count >= 8000
ORDER BY vote_count DESC;
```

### Exercise 5: Blockbusters By Two Directors (JOIN, IN)
Use a join to show movie `title`, director `name`, and `revenue` for films directed by `Peter Jackson` or `Tim Burton` with revenue at least `900000000`.
Order by `revenue` descending.


```python
%%sql
SELECT m.title, d.name AS director_name, m.revenue
FROM movies AS m
JOIN directors AS d ON m.director_id = d.id
WHERE d.name IN ('Peter Jackson', 'Tim Burton')
AND m.revenue >= 900000000
ORDER BY m.revenue DESC;
```

### Exercise 6: Film Counts For Named Directors (JOIN, GROUP BY, ORDER BY)
For these directors only: `Christopher Nolan`, `Steven Spielberg`, `Tim Burton`, `Ridley Scott`, `Michael Bay`, output director `name` and number of films as `movie_count`.
Group correctly and order by `movie_count` descending then `name` ascending.


```python
%%sql
SELECT d.name, COUNT(*) AS movie_count
FROM directors AS d
JOIN movies AS m ON d.id = m.director_id
WHERE d.name IN ('Christopher Nolan', 'Steven Spielberg', 'Tim Burton', 'Ridley Scott', 'Michael Bay')
GROUP BY d.id, d.name
ORDER BY movie_count DESC, d.name;
```

### Exercise 7: High-Rated Billion-Dollar Films (SELECT, WHERE, AND, ORDER BY)
Return `title` and `vote_average` for films with `vote_average` at least `8.0` and `revenue` at least `1000000000`.
Order by `vote_average` descending.


```python
%%sql
SELECT title, vote_average
FROM movies
WHERE vote_average >= 8.0 AND revenue >= 1000000000
ORDER BY vote_average DESC;
```
